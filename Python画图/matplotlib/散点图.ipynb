```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import seaborn as sns

# 导入数据
paper_name = np.array(['SimpleCIL', 'L2P', 'DualPrompt', 'ADAM w/ Adapter', 'Ours'])
paper_acc = np.array([85, 87, 89, 90, 96])
paper_params = np.array([0.1, 0.6, 0.7, 0.8, 0.41])
paper_time = np.array([0.1, 0.6, 7, 0.8, 9])

# 预设图像各种信息
large = 12
med = 8
small = 6
params = {'axes.titlesize': large,  # 子图上的标题字体大小
          'legend.fontsize': med,  # 图例的字体大小
          'figure.figsize': (med, small),  # 画布大小
          'axes.labelsize': med,  # 标签的字体大小
          'xtick.labelsize': large,  # x轴标尺的字体大小
          'ytick.labelsize': large,  # y轴标尺的字体大小
          'figure.titlesize': large}  # 整个画布的标题字体大小
plt.rcParams.update(params)  # 设定各种默认属性
plt.style.use('seaborn-whitegrid')  # 设置整体风格
sns.set_style('white')  # 设置整体背景风格

for i, name in enumerate(paper_name):
    plt.scatter(paper_params[i], paper_acc[i], s=paper_time[i]*20,alpha=0.5, linewidths=.5)

# 装饰图像
plt.gca().set(xlim=(0.0, 1), ylim=(0, 100),)
plt.title('', fontsize=22)
plt.xlabel('Tunable Parameters (Millon)', fontsize=15)
plt.ylabel('Average Accuracy (%)', fontsize=15)
plt.show() # 显示图像
```
