```python
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np

# Data for the heatmap
data = np.array([
    [82.4, 82.2, 82.2, 82.4, 82.7],
    [82.4, 82.4, 82.5, 82.6, 82.9],
    [82.4, 82.5, 82.5, 82.4, 82.5],
    [82.2, 82.3, 82.3, 82.1, 82.0],
    [81.6, 81.5, 81.5, 81.7, 81.5]
])

# Projection dimensions and trade-off parameters
r_labels = [128, 64, 32, 16, 8]
alpha_labels = [0.5, 0.3, 0.1, 5e-2, 1e-2]

# Create the heatmap
plt.figure(figsize=(6, 5))
sns.heatmap(data, xticklabels=r_labels, yticklabels=alpha_labels, annot=True, fmt=".1f", cmap="Blues", cbar_kws={'label': 'Accuracy (%)'})

# Add labels and title
plt.xlabel('Projection dimension r')
plt.ylabel('Trade-off Parameter α')
plt.title('(a) Robustness of hyperparameters', loc='left')

# Save the figure

# plt.savefig('/mnt/data/robustness_hyperparameters.png', dpi=300)

# Show the plot
plt.show()
```
