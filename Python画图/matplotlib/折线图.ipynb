```python
import numpy as np  
import matplotlib.pyplot as plt  
  
# x轴坐标，从50到100包含10个数值  
x = np.linspace(50, 100, 10)  
  
# y轴坐标，从60到100包含10个数值  
y = np.linspace(60, 100, 10)  
  
# 绘制折线图  
plt.plot(x, y, marker='o',markersize=4)  # marker='o' 表示在每个数据点上画一个圆圈  
  
# 设置轴标签和图表标题  
plt.xlabel('X-axis')  
plt.ylabel('Y-axis')  
plt.title('Line Plot with 10 Points')  
  
plt.figure(dpi=300, figsize=(12, 9))  # 设置图表分辨率和尺寸
plt.subplots_adjust(hspace=0.4, wspace=0.4)
# 显示网格  
# plt.grid(True)  
  
# 显示图表  
plt.show()
```
