```python
import matplotlib.pyplot as plt
import numpy as np

# Data
x = [1e0, 1e0, 1e1, 1e1, 1e2, 1e2, 1e2]
y = [76, 74, 78, 74, 76, 78, 72]
labels = ['L2P [CVPR\'22]', 'DualPrompt [ECCV\'22]', 'CODA-Prompt [CVPR\'23]', 'MEMO [ICLR\'23]', 'FOSTER [ECCV\'22]', 'DER [CVPR\'21]', 'Ours']
colors = ['orange', 'green', 'red', 'yellow', 'cyan', 'blue', 'red']
sizes = [100, 100, 100, 100, 100, 100, 300]

# Plot
plt.figure(figsize=(8, 6))
plt.scatter(x, y, s=sizes, c=colors, alpha=0.8)

# Adding labels
for i, label in enumerate(labels):
    if label == 'Ours':
        plt.text(x[i], y[i] + 0.5, label, ha='center', fontsize=12, weight='bold', color='red')
    else:
        plt.text(x[i], y[i] - 1.5, label, ha='center', fontsize=10)

# Customizing the plot
plt.xscale('log')
plt.xlim(0.5, 200)
plt.ylim(70, 80)
plt.xlabel('Trainable Parameters (%)', fontsize=12)
plt.ylabel('Average Accuracy (%)', fontsize=12)
plt.axvline(x=10, color='gray', linestyle='--')
plt.text(15, 71, 'Using Exemplars', fontsize=10, color='black')
plt.title('')
plt.grid(True, which="both", ls="--", linewidth=0.5)

# Display the plot
plt.show()
```
