```python
import matplotlib.pyplot as plt
import numpy as np  
  
# 假设的Expert 得分数据  
labels = ['Inc 0', 'Inc 1', 'Inc 2', 'Inc 3', 'Inc 4']  
scores = [[2000], [2174,1826], [1804,2389,1807], [2013,2977,2400,610], [1483,3631,2091,1474,1321]]
  
# 创建一个图形和一个子图  
fig, ax = plt.subplots()  
  
# 设置柱状图的宽度  
barWidth = 0.2  
  
# 设置柱状图的位置  
r1 = np.arange(len(labels))  
r2 = [x + barWidth for x in r1]  
r3 = [x + barWidth for x in r2]  
r4 = [x + barWidth for x in r3]  
  
# 绘制柱状图  
ax.bar(r1, scores[0], color='blue', width=barWidth, edgecolor='grey', label='Expert 1')  
ax.bar(r2, scores[1], color='green', width=barWidth, edgecolor='grey', label='Expert 2')  
ax.bar(r3, scores[2], color='red', width=barWidth, edgecolor='grey', label='Expert 0')  
ax.bar(r4, scores[3], color='purple', width=barWidth, edgecolor='grey', label='Expert 3')  
  
# 添加一些文本：标题、自定义x轴标签和y轴标签  
plt.xlabel('Inc', fontweight='bold')  
plt.ylabel('Expert Num', fontweight='bold')  
plt.xticks([r + barWidth for r in range(len(labels))], labels, rotation=45)  
  
# 添加图例  
plt.legend()  
  
# 显示图形  
plt.show()
```

```python
import matplotlib.pyplot as plt  
import numpy as np  
  
# 设置随机种子以便结果可重复  
np.random.seed(19680801)  
  
# 生成数据  
num_groups = 4  # 子图数量  
num_data_points = 200  # 总数据点数  
data = [np.random.randn(num_data_points) for _ in range(num_groups)]  
  
# 初始化图形和子图  
fig, axs = plt.subplots(nrows=2, ncols=2, figsize=(12, 8))  
axs = axs.flatten()  # 将二维数组转换为一维数组，方便遍历  
  
# 绘制柱状图  
bar_width = 1  
index = np.arange(num_data_points)  
group_size = 50  
  
for i, ax in enumerate(axs):  
    # 对数据进行分组  
    grouped_data = [data[i][j:j+group_size].mean() for j in range(0, num_data_points, group_size)]  
    x_pos = [j * group_size + group_size // 2 for j in range(len(grouped_data))]  
  
    # 绘制每组数据的平均值的柱状图  
    ax.bar(x_pos, grouped_data, bar_width, color='skyblue', align='center')  
  
    # 在每组结束处显示数据（这里显示的是平均值）  
    for x, val in zip(x_pos, grouped_data):  
        ax.annotate(f'{val:.2f}', (x, val + 0.5), textcoords="offset points", ha='center', va='bottom')  
  
    # 设置子图标题  
    ax.set_title(f'Model {i+1}')  
    ax.set_xticks(x_pos)  
    ax.set_xticklabels([f'{j*group_size}' for j in range(1, len(grouped_data)+1)])  
    ax.set_xlim(0, num_data_points)  
    ax.set_ylim(bottom=0)  
  
# 调整布局以防止子图重叠  
plt.tight_layout()  
  
# 显示图形  
plt.show()
```
